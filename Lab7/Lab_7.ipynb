$$
f(x,y) = \frac{1}{5} x^{2} + \frac{1}{5} y^{2}
$$
$$
\frac{\partial f}{\partial x}(x,y) = \frac{\partial f}{\partial x} = \frac{2}{5}x
$$
$$
\frac{\partial f}{\partial y}(x,y) = \frac{\partial f}{\partial y} = \frac{2}{5}y
$$
$$
A = (2, 1)
$$
$$
g(x, y) = \frac{\partial f}{\partial x}(x_{1}, y_{1})x + f(x_{1}, y_{1}) - \frac{\partial f}{\partial x}(x_{1}, y_{1})x_{1} = 
\frac{\partial f}{\partial x}(2, 1)x + f(2, 1) - \frac{\partial f}{\partial x}(2, 1)2 = 
\frac{4}{5}x + \frac{4}{5} + \frac{1}{5} - \frac{8}{5} = \frac{4}{5}x - \frac{3}{5}
$$
$$
h(x, y) = \frac{\partial f}{\partial y}(x_{1}, y_{1})x + f(x_{1}, y_{1}) - \frac{\partial f}{\partial y}(x_{1}, y_{1})x_{1} = \frac{\partial f}{\partial y}(2, 1)x + f(2, 1) - \frac{\partial f}{\partial y}(2, 1)2 = \frac{2}{5}x + \frac{4}{5} + \frac{1}{5} - \frac{4}{5} = \frac{2}{5}x + \frac{1}{5}
$$
