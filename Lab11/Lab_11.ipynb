```python
import matplotlib.pyplot as plt
import numpy as np
from typing import Callable

def draw_parametric(I: np.array, x: Callable[[np.ndarray], np.ndarray], y: Callable[[np.ndarray], np.ndarray]):
	t = np.linspace(I[0], I[1], 1000)
	X = x(t)
	Y = y(t)

	plt.figure(figsize=(6, 6))
	plt.plot(X, Y)
	
	max_range = max(
    np.max(X) - np.min(X),
    np.max(Y) - np.min(Y)
	) / 2
	x_mid = (np.max(X) + np.min(X)) / 2
	y_mid = (np.max(Y) + np.min(Y)) / 2
	plt.xlim(x_mid - max_range, x_mid + max_range)
	plt.ylim(y_mid - max_range, y_mid + max_range)
	
	plt.xlabel("x(t)")
	plt.ylabel("y(t)")
	plt.grid(True)
	plt.show()

# draw_parametric(
# 	[0, 2*np.pi],
# 	lambda t: np.cos(t),
# 	lambda t: np.sin(t)
# )
```

## Krzywe i ich parametryzacje

__Definicja.__ Krzywą $\gamma$ w przestrzeni  $\mathbb{R}^{n}$ nazywamy dowolny ciągły obraz odcinka $I=[a,b]$. Funkcję ciągłą $c\colon I\to\mathbb{R}^n$ nazywamy parametryzacją krzywej $\gamma$, o ile $\gamma = c(I)$.

### Przykłady: parametryzowanie odcinka

1. Funkcja $c:[0,2]\to\mathbb{R}^2$ o wzorze
$$
c(t)=(t,2t)
$$
jest parametryzacją odcinka łączącego punkt $(0,0)$ z punktem $(2,4)$.

2. Funkcja $c:[0,1]\to\mathbb{R}^2$
   o wzorze
$$
c(t)=(2t,4t) 
$$
jest parametryzacją odcinka łączącego punkt $(0,0)$ z punktem $(2,4)$.

3. Funkcja $c:[-1,3]\to\mathbb{R}^2$ o wzorze
$$
c(t)=(-t,t+1)
$$
jest parametryzacją odcinka łączącego punkt $(1,0)$ z punktem $(-3,4)$.


__Uwaga:__ przykłady 1 oraz 2 pokazują,  że dana krzywa ma wiele różnych parametryzacji.

#### Zadanie domowe 1
Podaj przykład parametryzacji odcinka łączącego:

a) punkt $(0,4)$ z punktem $(0,7)$; 

b) punkt $(1,1)$ z punktem $(-3,1)$;

c) punkt $(2,1)$ z punktem $(4,3)$.

a)
$$
c(t) = (0, -t + 7) \land I = [0, 3] \implies c(0) = (0, 7) \land c(3) = (0, 4)
$$

b)
$$
c(t) = (-3 + t, 1) \land I = [0, 4] \implies c(0) = (-3, 1) \land c(4) = (1, 1)
$$

```python
draw_parametric(
	[0, 3],
	lambda t: np.sin(2*np.pi * t/3),
	lambda t: -t + 7 + np.sin(2*np.pi * t/3)
)

draw_parametric(
	[0, 4],
	lambda t: -3 + t + np.sin(2*np.pi * t/3),
	lambda t: 1 + np.sin(2*np.pi/3 * t/3)
)
```

## Długość krzywej

__Definicja.__ Niech $$\gamma:[a, b]\to\mathbb{R}^2,\quad \gamma(t)=(\gamma_1(t),\gamma_2(t))$$ będzie krzywą gładką (to oznacza z definicji, że 
funkcje $\gamma_1$ oraz $\gamma_2$ mają ciągłe pochodne). Długość krzywej $\gamma$ oznaczamy
przez $L(\gamma)$ i definiujemy wzorem:

$$
L(\gamma) = \int_{a}^{b} |\gamma'(t)| dt=\int_a^b\sqrt{(\gamma_1'(t))^2+(\gamma_2'(t))^2}dt. 
$$

#### Zadanie domowe 4

Oblicz długość poniższych krzywych:

a) $\gamma:[0, 2]\to\mathbb{R}^2$, $\gamma(t)=(t,2-t)$;

b)  $\gamma:[0, 2\pi]\to\mathbb{R}^2$, $\gamma(t)=(3\cos t ,3\sin t)$.

a)
$$
\int_{0}^{2}\sqrt{((t)')^{2} + ((2 - t)')^{2}}dt = \int_{0}^{2}\sqrt{1^{2} + (-1)^{2}}dt = \int_{0}^{2}\sqrt{1 + 1}dt = \int_{0}^{2}\sqrt{2}dt = [t\sqrt{2}]_{0}^{2} = 2\sqrt{2}
$$

b)
$$
\int_{0}^{2\pi}\sqrt{((3\cos(t))')^{2} + ((3\sin(t))')^{2}}dt = \int_{0}^{2\pi}\sqrt{(-3\sin(t))^{2} + (3\cos(t))^{2}}dt = \int_{0}^{2\pi}\sqrt{9\sin(t)^{2} + 9\cos(t)^{2}}dt = \int_{0}^{2\pi}3\sqrt{\sin(t)^{2} + \cos(t)^{2}}dt = \int_{0}^{2\pi}3\sqrt{1}dt = \int_{0}^{2\pi}3dt = [3t]_{0}^{2\pi} = 6\pi
$$

```python
import sympy as sp


print("a)")
t = sp.symbols('t')

expr = sp.sqrt(sp.diff((t), t)**2 + sp.diff((2 - t), t)**2)
print(sp.integrate(expr, (t, 0, 2)))
print()

print("b)")
t = sp.symbols('t')

expr = sp.sqrt(sp.diff((3*sp.cos(t)), t)**2 + sp.diff((3*sp.sin(t)), t)**2)
print(sp.integrate(expr, (t, 0, 2 * sp.pi)))
```

## Całka krzywoliniowa niezorientowana (nieskierowana)

__Definicja__
Niech $$\gamma:[a, b]\to\mathbb{R}^2,\quad \gamma(t)=(\gamma_1(t),\gamma_2(t))$$ będzie krzywą gładką. Niech funkcja $f$ będzie funkcją ciągłą określoną na krzywej $\gamma$ (to jest na zbiorze $\gamma([a,b])$).

Liczbę 

$$
\int_a^bf(\gamma(t))\sqrt{(\gamma_1'(t))^2+(\gamma_2'(t))^2}dt
$$
nazywamy całką krzywoliniową niezorientowaną (nieskierowaną) z funkcji $f$ po  krzywej $\gamma$
i oznaczamy symbolem

$$
\int_\gamma fds.
$$

__Skąd nazwa?__ 

Jeśli po krzywej będziemy się poruszali od końca do początku, to wartość całki się  nie zmieni.

__Interpretacja fizyczna__

__1.__ Jeśli niejednorodny pręt o kształcie $\gamma$ ma gęstość $f$, to $\int_\gamma fds$ to jego masa.

__2.__ Jeśli funkcja $f$ jest nieujemna, to pole trapezu krzywoliniowego 



jest równe 
$\int_\gamma fds$.


### Przykład

Jeśli $\gamma:[0,1]\to\mathbb{R}^2$, $\gamma(t)=(3t,4t+1)$ a funkcja $f$ dana jest wzorem
$f(x,y)=x^2+y$, to 

$$
\int_\gamma fds=\int_0^1f(\gamma(t))\sqrt{(\gamma_1'(t))^2+(\gamma_2'(t))^2}dt=
\int_0^1f(3t,4t+1)\sqrt{3^2+4^2}dt=5\int_0^1(3t)^2+4t+1dt=30.
$$

## Całka krzywoliniowa zorientowana (skierowana)

__Przypomnienie:__

Iloczynem skalarnym wektorów $\overrightarrow{a}=(a_1,a_2)$ oraz $\overrightarrow{b}=(b_1,b_2)$ nazywamy liczbę
$$
\overrightarrow{a}\cdot\overrightarrow{b}=a_1b_1+a_2b_2.
$$

__Interpretacja fizyczna:__

Jeśli $\overrightarrow{F}$ jest wektorem siły a $\overrightarrow{d}$ jest wektorem przesunięcia, to liczba $\overrightarrow{F}\cdot \overrightarrow{d}$
jest równa pracy wykonanej przy przesunięciu punktu materialnego o wektor $\overrightarrow{d}$ z użyciem siły $\overrightarrow{F}$.

__Definicja całki krzywoliniowej zorientowanej__

Niech $$\gamma:[a, b]\to\mathbb{R}^2,\quad \gamma(t)=(\gamma_1(t),\gamma_2(t))$$ będzie krzywą gładką.  Niech funkcja $F(x,y)=(P(x,y),Q(x,y))$ będzie funkcją ciągłą (o wartościach w $\mathbb{R}^2$) określoną na krzywej $\gamma$ (to jest na zbiorze $\gamma([a,b])$).
Liczbę

$$
\int_a^bP(\gamma_1(t),\gamma_2(t))\gamma_1'(t)+Q(\gamma_1(t),\gamma_2(t))\gamma_2'(t)dt
$$
nazywamy całką krzywoliniową zorientowaną z pola wektorowego $F$ wzdłuż krzywej $\gamma$
i 
oznaczamy symbolem

$$
\int_\gamma P(x,y)dx+Q(x,y)dy.
$$
__Obserwacja:__
pod całką jest iloczyn skalarny wektorów $F(x,y)=(P(x,y),Q(x,y))$ i $(\gamma_1'(t),\gamma_2'(t))$.

__Terminologia:__ funkcję $F$ nazywamy polem wektorowym.

__Interpretacja fizyczna:__ całka krzywoliniowa, to praca jaką wykonujemy przesuwając punkt materialny wzdłuż krzywej $\gamma$ w polu elektrycznym lub grawitacyjnym $F$.

### Przykład

Niech $\gamma:[0,1]\to\mathbb{R}^2$ będzie krzywą o wzorze $\gamma(t)=(t^2,t^3)$.

1. Obliczymy

$$
\int_\gamma xdx+ydy.
$$
__UWAGA!!!!:__  to jest banalne zadanie, po prostu trzeba podstawić do wzoru.

Mamy:
$$
\int_\gamma xdx+ydy=\int_0^1t^2\cdot 2t+t^3\cdot3t^2dt=\ldots=1.
$$

2. Obliczymy

$$
\int_\gamma x+ydx.
$$

__UWAGA:__

To co mamy policzyć, to skrócona forma od zapisu: 
$$
\int_\gamma x+ydx+0dy.
$$
Mamy
$$
\int_\gamma x+ydx=\int_0^1(t^2+t^3)2tdt=\frac{9}{10}.
$$
